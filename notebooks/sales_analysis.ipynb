```python
total_sales = df_clean['Sales'].sum()
total_profit = df_clean['Profit'].sum()
total_quantity = df_clean['Quantity'].sum()
average_discount = df_clean['Discount'].mean()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
print("Average Discount:", round(average_discount, 2))
```

```python
profitable_records = (df_clean['Profit'] > 0).sum()
loss_records = (df_clean['Profit'] < 0).sum()
zero_profit_records = (df_clean['Profit'] == 0).sum()

print("Profitable Records:", profitable_records)
print("Loss Records:", loss_records)
print("Zero Profit Records:", zero_profit_records)
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df_clean['Sales'], bins=30)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(df_clean['Profit'], bins=40, kde=True)

plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    df_clean['Sales'],
    df_clean['Profit'],
    alpha=0.4
)

plt.title("Sales vs Profit Analysis")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
categorical_columns = df_clean.select_dtypes(
    include='object'
).columns.tolist()

print("Categorical Columns:")
print(categorical_columns)
```

```python
if 'Category' in df_clean.columns:
    category_sales = df_clean.groupby('Category')['Sales'].sum().sort_values(ascending=False)

    print(category_sales)

    category_sales.plot(
        kind='bar',
        figsize=(8, 5),
        title='Sales by Product Category',
        xlabel='Category',
        ylabel='Total Sales'
    )

    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
else:
    print("Category column not found.")
```

```python
if 'Region' in df_clean.columns:
    region_profit = df_clean.groupby('Region')['Profit'].sum().sort_values(ascending=False)

    print(region_profit)

    region_profit.plot(
        kind='bar',
        figsize=(8, 5),
        title='Profit by Region',
        xlabel='Region',
        ylabel='Total Profit'
    )

    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
else:
    print("Region column not found.")
```

```python
if 'City' in df_clean.columns:
    city_sales = (
        df_clean.groupby('City')['Sales']
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    print(city_sales)

    city_sales.plot(
        kind='bar',
        figsize=(11, 5),
        title='Top 10 Cities by Sales'
    )

    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
else:
    print("City column not found.")
```

```python
if 'Segment' in df_clean.columns:
    segment_sales = df_clean.groupby('Segment')['Sales'].sum()

    print(segment_sales)

    segment_sales.plot(
        kind='pie',
        autopct='%1.1f%%',
        figsize=(7, 7),
        title='Sales Distribution by Customer Segment'
    )

    plt.ylabel('')
    plt.show()
else:
    print("Segment column not found.")
```

```python
print("BUSINESS INSIGHTS")
print("-------------------------")

if 'Category' in df_clean.columns:
    print("Highest Sales Category:",
          category_sales.idxmax())

if 'Region' in df_clean.columns:
    print("Most Profitable Region:",
          region_profit.idxmax())

if 'City' in df_clean.columns:
    print("Top City by Sales:",
          city_sales.idxmax())

if 'Segment' in df_clean.columns:
    print("Top Customer Segment:",
          segment_sales.idxmax())
```

```python
discount_profit = df_clean.groupby('Discount')['Profit'].mean()

print(discount_profit)

plt.figure(figsize=(9, 5))
discount_profit.plot(kind='line', marker='o')

plt.title('Impact of Discount on Average Profit')
plt.xlabel('Discount')
plt.ylabel('Average Profit')
plt.grid(True)
plt.show()
```

```python
if 'Product Name' in df_clean.columns:

    product_profit = (
        df_clean.groupby('Product Name')['Profit']
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    print(product_profit)

    product_profit.plot(
        kind='barh',
        figsize=(10, 6),
        title='Top 10 Most Profitable Products'
    )

    plt.xlabel('Total Profit')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()

else:
    print("Product Name column not found.")
```

```python
if 'Product Name' in df_clean.columns:

    least_profitable = (
        df_clean.groupby('Product Name')['Profit']
        .sum()
        .sort_values()
        .head(10)
    )

    print(least_profitable)

    least_profitable.plot(
        kind='barh',
        figsize=(10, 6),
        title='Bottom 10 Products by Profit'
    )

    plt.xlabel('Total Profit')
    plt.tight_layout()
    plt.show()
```

```python
if 'Category' in df_clean.columns:

    category_quantity = (
        df_clean.groupby('Category')['Quantity']
        .sum()
        .sort_values(ascending=False)
    )

    print(category_quantity)

    category_quantity.plot(
        kind='bar',
        figsize=(8, 5),
        title='Quantity Sold by Category'
    )

    plt.ylabel('Total Quantity')
    plt.xticks(rotation=0)
    plt.show()
```

```python
total_sales = df_clean['Sales'].sum()
total_profit = df_clean['Profit'].sum()

profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Overall Profit Margin:", round(profit_margin, 2), "%")
```
